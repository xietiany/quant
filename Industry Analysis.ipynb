```python
potential_list = {}
runned = set()
```

```python
### search for stock
# industry = "半导体"
stock_board_industry_cons_em_df = ak.stock_board_industry_cons_em(symbol=industry)
excludeList = ['688428', '688520', '688235', '603014', '688351', '600930', '839493', '688729', '688234', '688256', '301678', '688981', '688512'\
              ,'001400', '301616']
i = 0
for item in stock_board_industry_cons_em_df.iterrows():
    # print(each[1])
    # print(each[1]["代码"], each[1]["名称"], each[1]["昨收"])
    each = item[1]
    if each["代码"][0] == "4" or each["代码"][0] == "8":
        ticker = "bj" + each["代码"]
    elif each["代码"][0] == "0" or each["代码"][0] == "3":
        ticker = "sz" + each["代码"]
    elif each["代码"][0] == "6":
        ticker = "sh" + each["代码"]
    print(ticker)
    if ticker in runned:
        continue
    else:
        runned.add(ticker)
    if each["市盈率-动态"] > 0 and each["代码"] not in excludeList:
        # print(ticker)
        try:
            a = stock(ticker)
            a.initialize(defaultRateApproach = True, valuationMethod = "earning", defaultLTGrowth = False, \
                                valuationStage = "three", growthCalcMethod = "earning", growthCalcHorizon = 5, valuationHorizon = 5, date='2024-09-15')
        except ValueError:
            print("there is a value error", ValueError)
        if a.firstStageGrowthValue:
            print("----------the current price for ", ticker, " is ", each["昨收"], "-------------------")
            print("----------the fair value for ", ticker, " is ", a.FV, "-------------------")
            if a.FV > each["昨收"]:
                potential_list[ticker] = {}
                potential_list[ticker]["代码"] = ticker
                potential_list[ticker]["公司名称"] = each["名称"]
                potential_list[ticker]["内在价值"] = a.FV
                potential_list[ticker]["现价"] = each["昨收"]
                potential_list[ticker]["行业"] = industry
        else:
            print("first stage growth is none")
    else:
        print("div not able to pull")
        print("PE is negative which means the stock doesn't have positive earning", each["市盈率-动态"])
    i += 1
    # if i == 10:
    #     break
```

```python
res = pd.DataFrame.from_dict(potential_list, orient='index')
res.index.name = '代码'
res
```

```python
res.to_csv(f"{industry}_v1.csv")
```
